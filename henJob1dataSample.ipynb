```python
import os

# Create an incoming directory inside your portfolio folder
incoming_dir = os.path.join(WORKSPACE_DIR, "incoming_customer_tickets")
os.makedirs(incoming_dir, exist_ok=True)

# A sample batch of incoming customer support issues
sample_tickets = {
    "ticket_001.txt": "I was charged twice on my credit card for my monthly subscription. Please refund the duplicate transaction.",
    "ticket_002.txt": "The mobile app keeps crashing every time I try to click the checkout button. I am using an iPhone 15.",
    "ticket_003.txt": "I forgot my password and the password reset link sent to my email address keeps returning a 404 expired page error.",
    "ticket_004.txt": "Can I update my billing address on my profile? I am moving to a different state next week.",
    "ticket_005.txt": "The video player on your website is lagging and won't load the high-definition stream. Audio works but image is frozen.",
    "ticket_006.txt": "My account has been locked due to too many failed login attempts. Can an administrator unlock it for me?",
    "ticket_007.txt": "I want to cancel my auto-renewal premium membership plan before the next billing cycle hits on the 15th.",
    "ticket_008.txt": "Your database API is throwing a 500 internal server error when trying to fetch user history tokens.",
    "ticket_009.txt": "I am receiving a suspicious notification saying a new device logged into my account from a location I don't recognize.",
    "ticket_010.txt": "Where can I download the annual tax invoice statements for my corporate account expenses?"
}

# Write out the simulated incoming files to disk
for filename, content in sample_tickets.items():
    file_path = os.path.join(incoming_dir, filename)
    with open(file_path, "w") as f:
        f.write(content)

print(f"Done! 10 sample incoming text files successfully written to: {incoming_dir}")
```

```python
import os
import glob
from transformers import pipeline

print("Loading classification model into memory (this may take a minute)...")
# Initialize a lightweight, high-performance classification pipeline
classifier = pipeline("zero-shot-classification", model="typeform/distilbert-base-uncased-mnli")

# Define target data pipeline criteria
TARGET_LABELS = ["BILLING", "TECHNICAL", "ACCOUNT_SECURITY"]
incoming_dir = os.path.join(WORKSPACE_DIR, "incoming_customer_tickets")

# Find all incoming text files in the target directory
file_pattern = os.path.join(incoming_dir, "*.txt")
target_files = glob.glob(file_pattern)

print(f"\n--- Starting Automated Pipeline Execution on {len(target_files)} Files ---\n")

# Process each file automatically
for file_path in sorted(target_files):
    filename = os.path.basename(file_path)

    # 1. Read the text file
    with open(file_path, "r") as f:
        ticket_text = f.read()

    # 2. Run the prompt/classification logic
    result = classifier(ticket_text, candidate_labels=TARGET_LABELS)

    # 3. Extract the highest confidence result
    top_prediction = result['labels'][0]
    confidence_score = result['scores'][0]

    # Print the automated routing action to log performance
    print(f"📄 File: {filename}")
    print(f"   ↳ Snippet: \"{ticket_text[:60]}...\"")
    print(f"   ⚡ Action: Route to [{top_prediction}] Queue (Confidence: {confidence_score:.2%})")
    print("-" * 70)

print("\nPipeline run complete! All files parsed, evaluated, and classified successfully.")
```

```python
import os
import glob
import shutil
from transformers import pipeline

# 1. Setup the concrete Directory Structure
MAIN_FOLDER_NAME = "automated_ticket_router"
TOP_LEVEL_DIR = os.path.join(WORKSPACE_DIR, MAIN_FOLDER_NAME)
INCOMING_DIR = os.path.join(TOP_LEVEL_DIR, "incoming_customer_tickets")
CLASSIFIED_DIR = os.path.join(TOP_LEVEL_DIR, "classified_storage")

# Ensure directories exist
os.makedirs(INCOMING_DIR, exist_ok=True)
os.makedirs(CLASSIFIED_DIR, exist_ok=True)

# 2. Migrate your original sample files into this clean architecture
original_incoming = os.path.join(WORKSPACE_DIR, "incoming_customer_tickets")
if os.path.exists(original_incoming):
    for original_file in glob.glob(os.path.join(original_incoming, "*.txt")):
        target_dest = os.path.join(INCOMING_DIR, os.path.basename(original_file))
        if not os.path.exists(target_dest):
            shutil.copy(original_file, target_dest)
    print("Simulated files successfully synchronized to new production directory structure.")

# 3. Initialize Classifier Configuration
print("Verifying pipeline model status...")
classifier = pipeline("zero-shot-classification", model="typeform/distilbert-base-uncased-mnli")
TARGET_LABELS = ["BILLING", "TECHNICAL", "ACCOUNT_SECURITY"]

# 4. Scan and Run the Processing Pipeline
target_files = glob.glob(os.path.join(INCOMING_DIR, "*.txt"))
print(f"\n--- Commencing File Processing and Sorting Layer on {len(target_files)} Files ---\n")

for file_path in target_files:
    filename = os.path.basename(file_path)

    # Read raw unstructured data
    with open(file_path, "r") as f:
        ticket_text = f.read()

    # Run model classification inference
    inference_result = classifier(ticket_text, candidate_labels=TARGET_LABELS)
    top_label = inference_result['labels'][0]  # Grab the highest confidence classification

    # Convert label to lowercase folder name (e.g., "ACCOUNT_SECURITY" -> "account_security")
    subfolder_name = top_label.lower()
    destination_subfolder = os.path.join(CLASSIFIED_DIR, subfolder_name)

    # Dynamically build out the unique subfolder if it hasn't been instantiated yet
    os.makedirs(destination_subfolder, exist_ok=True)

    # Execute the file system relocation move operation
    destination_file_path = os.path.join(destination_subfolder, filename)
    shutil.move(file_path, destination_file_path)

    print(f"📁 Processed: {filename}")
    print(f"   ↳ Text: \"{ticket_text[:50]}...\"")
    print(f"   🚚 Action: Moved to --> classified_storage/{subfolder_name}/")
    print("-" * 75)

print("\nProcessing complete! Check your Google Drive folder structure to see your sorted data lake.")
```

```python
print("--- Organized Storage Overview ---\n")
for root, dirs, files in os.walk(CLASSIFIED_DIR):
    level = root.replace(CLASSIFIED_DIR, '').count(os.sep)
    indent = ' ' * 4 * (level)
    print(f"{indent}📁 {os.path.basename(root)}/")
    subindent = ' ' * 4 * (level + 1)
    for f in sorted(files):
        print(f"{subindent}📄 {f}")
```
